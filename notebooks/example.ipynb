# PySpark cluster smoke test

This notebook submits a small DataFrame job to the Spark cluster in Docker Compose.

```python
!/opt/spark/bin/pyspark --version
```

```python
import os
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("cdc-streamer-notebook-example")
    .master(os.environ["SPARK_MASTER_URL"])
    .config("spark.driver.host", "jupyter")
    .config("spark.driver.bindAddress", "0.0.0.0")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")
```

```python
rows = [(1, "Raunak"), (2, "John"), (3, "Alice")]
people = spark.createDataFrame(rows, ["id", "name"])
people.show()
```

```python
spark.stop()
```
